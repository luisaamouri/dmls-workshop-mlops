````python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, ConfusionMatrixDisplay
import mlflow
````

````python
pd.set_option("display.max_columns", None)

# Load titanic dataset
titanic = sns.load_dataset("titanic")

print(f"ROWS, COLUMNS: {titanic.shape}")
titanic.head()
````

## **2. Data Preparation**

1. Create `pre-processing` Pipeline: 
- Impute (replace) missing numerical (median) and categorical (most frequent) \
    values in the preprocessing pipeline
- One-hot encode categoricals
- Standardize numeric features 
2. Split Train/Test split (80/20)

> You will need to remove or transform feature columns because of redundancy (e.g., `adult_male` is a column but we already have a `sex` column....`pclass` and `class`....`parch` and `sibsp` and `alone`....does it make sense to include `alive` if we're predicting `survived`?)

````python
# Count missing values
print(f"{titanic.isna().sum()}\n")
````

````python
# Summary of numeric columns 
titanic.describe().T
````

````python
# Unique values for categorical columns
for col in titanic.select_dtypes("object"):
    print(f"{col}: {titanic[col].unique()}")

# Summary of categorical columns 
titanic.describe(include="object").T
````

````python
# View class imbalance
titanic["survived"].value_counts()
````

````python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np

# Drop features 
redundant_cols = ["alive", "class", "alone", "adult_male", "embark_town"]
use_cols = [c for c in titanic.columns if c not in redundant_cols]

titanic_clean = titanic[use_cols].copy()

# Separate target
y = titanic_clean["survived"].astype(int)
X = titanic_clean.drop(columns=["survived"])

# Identify column types 
numeric_cols = X.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = X.select_dtypes(include=["object", "category"]).columns.tolist()

#  Preprocessing pipeline
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_cols),
        ("cat", categorical_transformer, categorical_cols),
    ],
    remainder="drop"
)

# 80/20 Train/Test split (stratified) 
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# Fit training data, transform test data
X_train_proc = preprocessor.fit_transform(X_train)
X_test_proc = preprocessor.transform(X_test)
````

## **3. Model Training & EDA**

1. Explore dataset through visualizations
- Check class imbalance with bar chart for `survived`
- Compare surival rate by `sex`, `class`, and `embarked`
- Plot histograms/boxplots of `age` and `fare` split by `survived`
- Show pairplot for correlations between numeric features

2. Build a `Pipeline` that chains the **preprocessor** 
    and a **baseline model** like `LogisticRegression` or `RandomForestClassifer`

3. Train on the 80 % train set and evaluate on the held‑out 20 % test set

4. Save metrics (i.e., accuracy, test_f1_macro, test_recall_macro, test_precision_macro) for logging


````python
# -----------CODE HERE-------------
from sklearn.metrics import accuracy_score, f1_score, recall_score, precision_score
import mlflow.sklearn

# Class imbalance 
sns.countplot(data=titanic, x="survived", palette="Set2", hue=None)
plt.title("Class Distribution: Survived vs Not Survived")
plt.show()

# Survival by sex 
sns.barplot(data=titanic, x="sex", y="survived", palette="Set2", hue=None)
plt.title("Survival Rate by Sex")
plt.show()

# Survival by class 
sns.barplot(data=titanic, x="pclass", y="survived", palette="Set2", hue=None)
plt.title("Survival Rate by Passenger Class")
plt.show()

# Survival by embarkment
sns.barplot(data=titanic, x="embarked", y="survived", palette="Set2", hue=None)
plt.title("Survival Rate by Embarked Location")
plt.show()

# Age distribution split by survival
sns.histplot(data=titanic, x="age", hue="survived", multiple="stack", palette="Set2", bins=30)
plt.title("Age Distribution by Survival")
plt.show()

# Fare distribution split by survival
sns.boxplot(data=titanic, x="survived", y="fare", palette="Set2")
plt.title("Fare Distribution by Survival")
plt.show()

# Pairplot for numeric correlations 
numeric_features = ["age", "fare", "sibsp", "parch", "pclass"]
sns.pairplot(titanic[numeric_features + ["survived"]], hue="survived", palette="Set2")
plt.suptitle("Pairwise Correlations between Numeric Features", y=1.02)
plt.show()

# Model selection
baseline_model = LogisticRegression(max_iter=1000, random_state=42)

# Pipeline
model_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", baseline_model)
])

# Fit model
model_pipeline.fit(X_train, y_train)
mlflow.sklearn.save_model(model_pipeline, "deliverable/api/artifacts/model")

# Predict on test set
y_pred = model_pipeline.predict(X_test)

# Calculate and save metrics
accuracy = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred, average="macro")
recall = recall_score(y_test, y_pred, average="macro")
precision = precision_score(y_test, y_pred, average="macro")

print(f"Accuracy: {accuracy:.4f}")
print(f"F1 (Macro): {f1:.4f}")
print(f"Recall (Macro): {recall:.4f}")
print(f"Precision (Macro): {precision:.4f}")
````

````python
# -----------CODE HERE-------------

from mlflow.tracking import MlflowClient
from sklearn.metrics import confusion_matrix
client = MlflowClient()
import time


# Set up tracking server and create second experiment 
mlflow.set_tracking_uri("http://127.0.0.1:5000")
mlflow.set_experiment("titanic_classification_model_family_comparison")

PRIMARY_METRIC = "test_f1_macro"
TIEBREAKERS = ["test_recall_macro", "test_accuracy"]
registered = []

client = MlflowClient()

model_name = "titanic-classifier"           # 
run_name = f"titanic-lr-baseline-{int(time.time())}"
feature_change = "baseline-preprocessor-v1" # 
params = {
    "model_family": "lr",
    "model": "LogisticRegression",
    "max_iter": model_pipeline.named_steps["classifier"].max_iter,
    "C": model_pipeline.named_steps["classifier"].C,
    "solver": model_pipeline.named_steps["classifier"].solver,
    "random_state": getattr(model_pipeline.named_steps["classifier"], "random_state", None),
}
metrics = {
    "test_accuracy": accuracy,
    "test_f1_macro": f1,
    "test_recall_macro": recall,
    "test_precision_macro": precision,
}
input_sample = X_test.head(5) 
cm = confusion_matrix(y_test, y_pred, labels=[0, 1])
cm_fig = sns.heatmap(cm, annot=True, fmt="d", cbar=False,
                     xticklabels=["pred_0", "pred_1"], yticklabels=["true_0", "true_1"]).get_figure()
cm_png = "confusion_matrix_titanic.png"
cm_fig.suptitle("Confusion Matrix - Titanic Logistic Regression")
cm_fig.savefig(cm_png, bbox_inches="tight")
plt.close(cm_fig)

model_family = "lr" if "lr" in model_name or "logistic" in model_name.lower() else "rf"

with mlflow.start_run(run_name=run_name) as run:
    run_id = run.info.run_id


    mlflow.set_tags({
        "model_family": model_family,
        "feature_change": feature_change,
        "dataset": "seaborn-titanic",
        "framework": "scikit-learn",
        "owner": "you", 
    })

    mlflow.log_params(params)
    mlflow.log_metrics(metrics)
    mlflow.log_artifact(cm_png, artifact_path="plots")

    model_info = mlflow.sklearn.log_model(
        sk_model=model_pipeline,
        artifact_path="model",
        input_example=input_sample,
        registered_model_name=model_name
    )

    mv = client.search_model_versions(
        f"name = '{model_name}' and run_id = '{run_id}'"
    )[0]
    version = mv.version

    client.set_model_version_tag(model_name, version, "model_family", model_family)
    client.set_model_version_tag(model_name, version, "feature_change", feature_change)
    client.set_model_version_tag(model_name, version, "run_id", run_id)
    client.set_model_version_tag(model_name, version, "run_name", run_name)

    selection_tuple = (
        metrics[PRIMARY_METRIC],
        metrics.get(TIEBREAKERS[0], -1),
        metrics.get(TIEBREAKERS[1], -1),
        model_name,
        version,
        run_name
    )
    registered.append(selection_tuple)

print("Logged & registered. Selection tuple:", registered[-1])
````
